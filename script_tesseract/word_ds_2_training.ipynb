```python
import os
import glob
import pytesseract
from PIL import Image
from jiwer import cer, wer

data_dir = r"\\wsl$\Ubuntu\home\janmejaywsl\tesstrain\data\guj-ground-truth"

# Define the paths as clean raw strings without extra internal quotes
old_model_dir = r"D:\code material 2.0\akshar-setu-v2\model\word\old"
new_model_dir = r"D:\code material 2.0\akshar-setu-v2\model\word\new"

ground_truths = []
old_predictions = []
new_predictions = []

print("Gathering image files...")
image_files = glob.glob(os.path.join(data_dir, "*.png"))
image_files = image_files[:200]

print(f"Starting evaluation on {len(image_files)} images...")

for idx, img_path in enumerate(image_files):
    gt_path = img_path.replace(".png", ".gt.txt")

    if not os.path.exists(gt_path):
        continue

    with open(gt_path, "r", encoding="utf-8") as f:
        true_text = f.read().strip()

    img = Image.open(img_path)
    ground_truths.append(true_text)

    # Switch environment to Old Model directory
    os.environ['TESSDATA_PREFIX'] = old_model_dir
    old_text = pytesseract.image_to_string(img, lang='guj').strip()
    old_predictions.append(old_text)

    # Switch environment to New Model directory
    os.environ['TESSDATA_PREFIX'] = new_model_dir
    new_text = pytesseract.image_to_string(img, lang='guj').strip()
    new_predictions.append(new_text)

    if (idx + 1) % 20 == 0:
        print(f"Processed {idx + 1}/{len(image_files)} images...")

# --- CALCULATE METRICS ---
old_cer = cer(ground_truths, old_predictions)
old_wer = wer(ground_truths, old_predictions)

new_cer = cer(ground_truths, new_predictions)
new_wer = wer(ground_truths, new_predictions)

print("\n--- RESULTS ---")
print(f"Old Model -> CER: {old_cer:.4f} | WER: {old_wer:.4f}")
print(f"New Model -> CER: {new_cer:.4f} | WER: {new_wer:.4f}")

cer_improvement = (old_cer - new_cer) / old_cer * 100 if old_cer > 0 else 0
print(f"\nCER Improvement: {cer_improvement:.2f}%")
```

```python
# --- CALCULATE METRICS ---
old_cer = cer(ground_truths, old_predictions)
old_wer = wer(ground_truths, old_predictions)

new_cer = cer(ground_truths, new_predictions)
new_wer = wer(ground_truths, new_predictions)

# Convert CER to traditional Accuracy Scores (Clamped at 0% minimum)
old_accuracy = max(0, (1 - old_cer) * 100)
new_accuracy = max(0, (1 - new_cer) * 100)

print("\n--- RESULTS ---")
print(f"Old Model -> Accuracy: {old_accuracy:.2f}% | CER: {old_cer:.4f} | WER: {old_wer:.4f}")
print(f"New Model -> Accuracy: {new_accuracy:.2f}% | CER: {new_cer:.4f} | WER: {new_wer:.4f}")

cer_improvement = (old_cer - new_cer) / old_cer * 100 if old_cer > 0 else 0
print(f"\nError Rate Improvement: {cer_improvement:.2f}%")
```
