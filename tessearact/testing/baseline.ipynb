```python
import os
import glob
import pytesseract
from PIL import Image
from jiwer import cer, wer


# ============================================================
# BASELINE MODEL
# ============================================================

baseline_model_dir = (
    r"D:\code material 2.0\akshar-setu-v2\model\baseline"
)

os.environ["TESSDATA_PREFIX"] = baseline_model_dir


# ============================================================
# DATASET PATHS
# ============================================================

# WORD DATASET
word_data_dir = (
    r"\\wsl$\Ubuntu\home\janmejaywsl\tesstrain\data\guj-ground-truth"
)

# LINE DATASET
line_data_dir = (
    r"\\wsl.localhost\Ubuntu\home\janmejaywsl\tesstrain\data\guj-experiment2-ground-truth"
)


# ============================================================
# EVALUATION FUNCTION
# ============================================================

def evaluate_dataset(data_dir, dataset_name):

    print("\n")
    print("=" * 70)
    print(f"Evaluating Baseline Model on {dataset_name}")
    print("=" * 70)

    # --------------------------------------------------------
    # Get images
    # --------------------------------------------------------

    image_files = glob.glob(
        os.path.join(data_dir, "*.png")
    )

    # Evaluate first 200 images
    image_files = image_files[:200]

    print(
        f"Images found: {len(image_files)}"
    )

    ground_truths = []
    predictions = []


    # --------------------------------------------------------
    # OCR
    # --------------------------------------------------------

    for idx, img_path in enumerate(image_files):

        gt_path = img_path.replace(
            ".png",
            ".gt.txt"
        )

        if not os.path.exists(gt_path):

            print(
                f"WARNING: Ground truth missing: "
                f"{os.path.basename(img_path)}"
            )

            continue


        # Read ground truth
        with open(
            gt_path,
            "r",
            encoding="utf-8"
        ) as f:

            true_text = f.read().strip()


        # Load image
        img = Image.open(img_path)


        # Baseline OCR
        predicted_text = pytesseract.image_to_string(
            img,
            lang="guj"
        ).strip()


        ground_truths.append(true_text)
        predictions.append(predicted_text)


        # Progress
        if (idx + 1) % 20 == 0:

            print(
                f"Processed "
                f"{idx + 1}/{len(image_files)}..."
            )


    # ========================================================
    # METRICS
    # ========================================================

    if len(ground_truths) == 0:

        print("No valid images found.")

        return None


    # CER
    character_error_rate = cer(
        ground_truths,
        predictions
    )


    # WER
    word_error_rate = wer(
        ground_truths,
        predictions
    )


    # Exact Match Accuracy
    exact_matches = sum(
        gt.strip() == pred.strip()
        for gt, pred in zip(
            ground_truths,
            predictions
        )
    )

    exact_match_accuracy = (
        exact_matches / len(ground_truths)
    )


    # ========================================================
    # RESULTS
    # ========================================================

    print("\n")
    print("-" * 70)
    print(f"{dataset_name} — BASELINE MODEL RESULTS")
    print("-" * 70)

    print(
        f"Images Evaluated      : {len(ground_truths)}"
    )

    print(
        f"CER                   : "
        f"{character_error_rate * 100:.2f}%"
    )

    print(
        f"WER                   : "
        f"{word_error_rate * 100:.2f}%"
    )

    print(
        f"Exact Match Accuracy  : "
        f"{exact_match_accuracy * 100:.2f}%"
    )

    print("-" * 70)


    return {
        "images": len(ground_truths),
        "CER": character_error_rate,
        "WER": word_error_rate,
        "Exact Match Accuracy": exact_match_accuracy
    }


# ============================================================
# 1. BASELINE ON WORD DATASET
# ============================================================

word_results = evaluate_dataset(
    word_data_dir,
    "WORD DATASET"
)


# ============================================================
# 2. BASELINE ON LINE DATASET
# ============================================================

line_results = evaluate_dataset(
    line_data_dir,
    "LINE DATASET"
)


# ============================================================
# FINAL COMPARISON
# ============================================================

print("\n")
print("=" * 80)
print("          BASELINE MODEL — DATASET COMPARISON")
print("=" * 80)

print(
    f"{'Metric':<28}"
    f"{'WORD DATASET':>22}"
    f"{'LINE DATASET':>22}"
)

print("-" * 80)


print(
    f"{'Images Evaluated':<28}"
    f"{word_results['images']:>22}"
    f"{line_results['images']:>22}"
)


print(
    f"{'CER':<28}"
    f"{word_results['CER'] * 100:>21.2f}%"
    f"{line_results['CER'] * 100:>21.2f}%"
)


print(
    f"{'WER':<28}"
    f"{word_results['WER'] * 100:>21.2f}%"
    f"{line_results['WER'] * 100:>21.2f}%"
)


print(
    f"{'Exact Match Accuracy':<28}"
    f"{word_results['Exact Match Accuracy'] * 100:>21.2f}%"
    f"{line_results['Exact Match Accuracy'] * 100:>21.2f}%"
)


print("=" * 80)
```
