```python
import os
import glob
import pytesseract
from PIL import Image
from jiwer import cer, wer

# ============================================================
# PATHS
# ============================================================

data_dir = r"D:\code material 2.0\akshar-setu-v3\eval_sets\99_percent_correct"

# OLD = BASELINE MODEL
old_model_dir = r"D:\code material 2.0\akshar-setu-v2\model\baseline"

# NEW = LINE DATASET MODEL
new_model_dir = r"D:\code material 2.0\akshar-setu-v2\model\line\syntheticline"


# ============================================================
# DATA STORAGE
# ============================================================

ground_truths = []

# Separate storage for each model
baseline_predictions = []
line_dataset_predictions = []


# ============================================================
# LOAD IMAGES
# ============================================================

print("Gathering image files...")

image_files = glob.glob(os.path.join(data_dir, "*.png"))

# Evaluate first 200 images
image_files = image_files[:200]

print(f"Starting evaluation on {len(image_files)} images...\n")


# ============================================================
# EVALUATION
# ============================================================

for idx, img_path in enumerate(image_files):

    gt_path = img_path.replace(".png", ".gt.txt")

    if not os.path.exists(gt_path):
        continue

    # -------------------------
    # Ground Truth
    # -------------------------

    with open(gt_path, "r", encoding="utf-8") as f:
        true_text = f.read().strip()

    img = Image.open(img_path)

    ground_truths.append(true_text)


    # ========================================================
    # MODEL 1 — BASELINE
    # ========================================================

    os.environ["TESSDATA_PREFIX"] = old_model_dir

    baseline_text = pytesseract.image_to_string(
        img,
        lang="guj"
    ).strip()

    baseline_predictions.append(baseline_text)


    # ========================================================
    # MODEL 2 — LINE DATASET MODEL
    # ========================================================

    os.environ["TESSDATA_PREFIX"] = new_model_dir

    line_dataset_text = pytesseract.image_to_string(
        img,
        lang="guj"
    ).strip()

    line_dataset_predictions.append(line_dataset_text)


    # Progress
    if (idx + 1) % 20 == 0:
        print(f"Processed {idx + 1}/{len(image_files)} images...")


# ============================================================
# METRIC FUNCTIONS
# ============================================================

def calculate_metrics(ground_truth, predictions):

    # CER
    character_error_rate = cer(
        ground_truth,
        predictions
    )

    # WER
    word_error_rate = wer(
        ground_truth,
        predictions
    )

    # Character Accuracy
    character_accuracy = 1 - character_error_rate

    # Word Accuracy
    word_accuracy = 1 - word_error_rate

    # Exact Match Accuracy
    exact_matches = sum(
        gt.strip() == pred.strip()
        for gt, pred in zip(ground_truth, predictions)
    )

    exact_match_accuracy = (
        exact_matches / len(ground_truth)
        if len(ground_truth) > 0
        else 0
    )

    return {
        "CER": character_error_rate,
        "WER": word_error_rate,
        "Character Accuracy": character_accuracy,
        "Word Accuracy": word_accuracy,
        "Exact Match Accuracy": exact_match_accuracy
    }


# ============================================================
# CALCULATE RESULTS
# ============================================================

baseline_results = calculate_metrics(
    ground_truths,
    baseline_predictions
)

line_dataset_results = calculate_metrics(
    ground_truths,
    line_dataset_predictions
)


# ============================================================
# IMPROVEMENT
# ============================================================

cer_improvement = (
    (baseline_results["CER"] - line_dataset_results["CER"])
    / baseline_results["CER"] * 100
    if baseline_results["CER"] > 0
    else 0
)

wer_improvement = (
    (baseline_results["WER"] - line_dataset_results["WER"])
    / baseline_results["WER"] * 100
    if baseline_results["WER"] > 0
    else 0
)


# ============================================================
# FINAL ACCURACY MATRIX
# ============================================================

print("\n")
print("=" * 80)
print("                 GUJARATI OCR MODEL EVALUATION")
print("=" * 80)

print(f"Dataset: {len(ground_truths)} images")
print()

print(
    f"{'Metric':<25}"
    f"{'BASELINE':>18}"
    f"{'LINE-DATASET MODEL':>25}"
)

print("-" * 80)

print(
    f"{'CER':<25}"
    f"{baseline_results['CER'] * 100:>17.2f}%"
    f"{line_dataset_results['CER'] * 100:>24.2f}%"
)

print(
    f"{'WER':<25}"
    f"{baseline_results['WER'] * 100:>17.2f}%"
    f"{line_dataset_results['WER'] * 100:>24.2f}%"
)

print(
    f"{'Character Accuracy':<25}"
    f"{baseline_results['Character Accuracy'] * 100:>17.2f}%"
    f"{line_dataset_results['Character Accuracy'] * 100:>24.2f}%"
)

print(
    f"{'Word Accuracy':<25}"
    f"{baseline_results['Word Accuracy'] * 100:>17.2f}%"
    f"{line_dataset_results['Word Accuracy'] * 100:>24.2f}%"
)

print(
    f"{'Exact Match Accuracy':<25}"
    f"{baseline_results['Exact Match Accuracy'] * 100:>17.2f}%"
    f"{line_dataset_results['Exact Match Accuracy'] * 100:>24.2f}%"
)

print("=" * 80)


# ============================================================
# IMPROVEMENT SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("                    IMPROVEMENT SUMMARY")
print("=" * 80)

print(f"Baseline CER          : {baseline_results['CER'] * 100:.2f}%")
print(f"Line Dataset CER      : {line_dataset_results['CER'] * 100:.2f}%")
print(f"CER Improvement       : {cer_improvement:.2f}%")

print()

print(f"Baseline WER          : {baseline_results['WER'] * 100:.2f}%")
print(f"Line Dataset WER      : {line_dataset_results['WER'] * 100:.2f}%")
print(f"WER Improvement       : {wer_improvement:.2f}%")

print("=" * 80)
```
